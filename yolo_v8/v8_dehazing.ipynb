```python
import cv2
import numpy as np

def dehaze(img):
    # Simple dehazing with histogram equalization
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv)
    v_eq = cv2.equalizeHist(v)
    hsv_eq = cv2.merge((h, s, v_eq))
    dehazed = cv2.cvtColor(hsv_eq, cv2.COLOR_HSV2BGR)
    return dehazed
```

```python
import os
from tqdm import tqdm
import matplotlib.pyplot as plt

def process_folder(input_path="runs/dataset/test/images", output_path="output_dehazed", preview=True, preview_count=3):
    if not os.path.exists(input_path):
        print(f"⚠️ Input folder not found: {input_path}")
        return
    
    os.makedirs(output_path, exist_ok=True)
    shown = 0
    
    for fname in tqdm(os.listdir(input_path)):
        fpath = os.path.join(input_path, fname)
        img = cv2.imread(fpath)
        if img is None:
            continue
        
        # Apply dehazing
        dehazed = dehaze(img)
        
        # Save result
        save_path = os.path.join(output_path, fname)
        success = cv2.imwrite(save_path, dehazed)
        if success:
            print(f"✅ Saved: {save_path}")
        else:
            print(f"❌ Failed to save: {save_path}")

        # Show preview
        if preview and shown < preview_count:
            before = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            after = cv2.cvtColor(dehazed, cv2.COLOR_BGR2RGB)

            plt.figure(figsize=(10, 5))
            plt.subplot(1, 2, 1)
            plt.imshow(before)
            plt.title("Before")
            plt.axis("off")

            plt.subplot(1, 2, 2)
            plt.imshow(after)
            plt.title("After")
            plt.axis("off")

            plt.show()

            shown += 1
```

```python
process_folder()
```
